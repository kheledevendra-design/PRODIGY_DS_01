```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings; warnings.filterwarnings('ignore')
```

```python
# 1. Load the dataset
# Skip the first 4 rows of metadata
df = pd.read_csv(r'C:\Users\Sumit\Downloads\API_SP.POP.TOTL_DS2_en_csv_v2_34\API_SP.POP.TOTL_DS2_en_csv_v2_34.csv', skiprows=4)
df.head()
```

```python
# 2. Data Cleaning
# Remove empty columns and choose the year 2023 for visualization
df_clean = df.drop(columns=['Unnamed: 69'])
pop_2023 = df_clean[['Country Name', '2023']].dropna()

# Set the style for the plots
sns.set(style="whitegrid")
```

```python
# 3. Create a Histogram (Distribution of Continuous Variable)
plt.figure(figsize=(10, 6))
sns.histplot(pop_2023['2023'], bins=30, kde=True, color='skyblue')
plt.title('Distribution of Total Population (2023)', fontsize=15)
plt.xlabel('Population (Log Scale)', fontsize=12)
plt.ylabel('Number of Countries/Regions', fontsize=12)
plt.xscale('log')  # Log scale used to accommodate the wide range of values
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
# 4. Create a Bar Chart (Top 10 Most Populous Entities)
# We sort the data to find the largest entities
top_10_pop = pop_2023.sort_values(by='2023', ascending=False).head(10)

plt.figure(figsize=(12, 8))
sns.barplot(x='2023', y='Country Name', data=top_10_pop, palette='viridis')
plt.title('Top 10 Most Populous Entities in 2023', fontsize=15)
plt.xlabel('Total Population', fontsize=12)
plt.ylabel('Country / Region Name', fontsize=12)
plt.tight_layout()
plt.show()
```
